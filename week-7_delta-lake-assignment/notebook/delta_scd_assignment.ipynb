## Step 2: Perform Basic Data Cleaning

In this step, the Delta table is cleaned by identifying null values, replacing missing data, and removing duplicate records to improve data quality.

```python
# Read the Delta table

customer_df = spark.table("customer_delta")

display(customer_df)
```

```python
# Count null values in each column

from pyspark.sql.functions import col, sum

customer_df.select([
    sum(col(c).isNull().cast("int")).alias(c)
    for c in customer_df.columns
]).show()
```

fill null values using median

```python
# Calculate median age
median_age = customer_df.approxQuantile("age", [0.5], 0)[0]

# Fill missing age with median
customer_df = customer_df.fillna({"age": median_age})
```

```python
customer_df = customer_df.fillna({
    "email": "unknown@mail.com"
})

# display the updated dataset
display(customer_df)
```

Replace missing email values with 'unknown@mail.com' because email cannot be inferred, so a placeholder is more appropriate.

## handle duplicates values

```python
# Count of total duplicate rows
print("Total Duplicate Rows:",customer_df.count() - customer_df.distinct().count())
```

there are total 27 duplicate rows hence perform removing of duplication.

```python
# Remove duplicate rows

customer_df = customer_df.dropDuplicates()
```

```python
# verify duplicates were removed

print("Total Duplicate Rows:",customer_df.count() - customer_df.distinct().count())
```

```python
customer_df = customer_df.dropDuplicates(["customer_id"])
```

# Saving the updated dataframe back into delta table

```python
# Save the cleaned data back to the Delta table

customer_df.write \
    .format("delta") \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable("customer_delta")
```

```python
# Read the updated Delta table

display(spark.table("customer_delta"))
```

## Step 3: Create Incremental Dataset

In this step, a new dataset is created to simulate incoming customer data. It contains existing customer records for updates and new customer records for insertion.

```python
# Read the cleaned Delta table

customer_df = spark.table("customer_delta")
```

### Create Update Records

Select existing customers and modify some column values to simulate updated records.

```python
# Create records to update existing customers

updates = (
    customer_df
    .filter(col("customer_id").between(4901, 5000))
    .withColumn("membership", lit("Platinum"))
    .withColumn("total_spent", col("total_spent") + 5000)
)

display(updates)
```

### Create 2nd Dataset Customer Records(customer_incremental.csv)

Generate new customer records that do not exist in the Delta table.

```python
# Create new customer records

new_customers = (
    spark.range(5001, 5101)
    .withColumnRenamed("id", "customer_id")
    .withColumn("customer_name", concat(lit("Customer_"), col("customer_id")))
    .withColumn("email", concat(lit("customer"), col("customer_id"), lit("@mail.com")))
    .withColumn("city", lit("Pune"))
    .withColumn("state", lit("MH"))
    .withColumn("age", lit(28))
    .withColumn("gender", lit("Male"))
    .withColumn("membership", lit("Silver"))
    .withColumn("last_purchase", current_date())
    .withColumn("total_spent", lit(1500.0))
)

display(new_customers)
```

### Create Incremental Dataset

Combine updated records and new customer records into a single incremental dataset.

```python
# Combine update and insert records

incremental_df = updates.union(new_customers)

display(incremental_df)
```

### Save Incremental Dataset

Store the incremental dataset as a CSV file to simulate incoming batch data.

```python
# Save incremental dataset as CSV

incremental_df.coalesce(1).write \
    .mode("overwrite") \
    .option("header", "true") \
    .csv("/Volumes/workspace/default/delta_volume/customer_incremental")
```

### Read Incremental Dataset

Load the generated incremental CSV file for performing the Delta Lake MERGE operation.

```python
# Read incremental CSV

incremental_df = (
    spark.read
    .format("csv")
    .option("header", "true")
    .option("inferSchema", "true")
    .load("/Volumes/workspace/default/delta_volume/customer_incremental/part-00000-tid-8381575887906411546-e13deea0-050e-43da-8fdb-f3b2a06eea6b-219-1-c000.csv")
)

# Display incremental dataset
display(incremental_df)
```

## Step 4: Create Incremental Delta Table and Apply MERGE

In this step, the incremental customer dataset is stored as a Delta table. The Delta MERGE operation is then used to update existing records and insert new records into the main customer Delta table.

### Create Incremental Delta Table

Store the incremental dataset as a Delta table for performing the MERGE operation.

```python
# load incremental dataset as Delta table
incremental_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("customer_incremental_delta")
```

### Verify Incremental Delta Table

Display the incremental Delta table to verify that it was created successfully.

```python
%sql
select * from customer_incremental_delta
```

### Load Delta Tables

Load both the customer Delta table and the incremental Delta table for the MERGE operation.

```python
# Load Delta tables

from delta.tables import DeltaTable

customer_delta = DeltaTable.forName(spark, "customer_delta")

incremental_delta = DeltaTable.forName(spark, "customer_incremental_delta")
```

### Merge Delta Tables

Update matching customer records and insert new customer records into the main Delta table.

```python
# Merge incremental Delta table into customer Delta table

(
    customer_delta.alias("target")
    .merge(
        incremental_delta.toDF().alias("source"),
        "target.customer_id = source.customer_id"
    )
    .whenMatchedUpdateAll()
    .whenNotMatchedInsertAll()
    .execute()
)
```

### Verify Merge Result

Display the updated customer Delta table after the MERGE operation.

```python
# Display updated customer Delta table

display(spark.table("customer_delta"))
```

## Step 5: Validate Results

In this step, the final Delta table is validated by checking the total number of records, identifying duplicate customer IDs, and verifying that the MERGE operation was successful.


### Check Total Records

Count the total number of records in the customer Delta table after the MERGE operation.

total record is now 5100 which means that it succeffully merge incremental_customer.csv

### Check Duplicate Customer IDs

Verify that no duplicate customer IDs exist in the Delta table.

```python
# Check duplicate customer IDs

duplicates = (
    spark.table("customer_delta")
    .groupBy("customer_id")
    .count()
    .filter("count > 1")
)

display(duplicates)
```

### Verify Updated Records

Display updated customer records to confirm that the MERGE operation updated existing records successfully.

```python
# Display updated customer records

display(
    spark.table("customer_delta")
    .filter("customer_id BETWEEN 4901 AND 5000").orderBy("customer_id")
)
```

### Verify New Records

Display newly inserted customer records to confirm that the MERGE operation inserted new records successfully.

```python
# Display newly inserted customer records

display(
    spark.table("customer_delta")
    .filter("customer_id >= 5001").orderBy("customer_id")
)
```
