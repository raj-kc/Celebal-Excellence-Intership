# Step 2: Create Schemas under shadow_revenue_catalog

### Description

The project follows the Medallion Architecture, which separates data into three logical layers.

- Bronze: Stores raw data exactly as received.
- Silver: Stores cleaned and validated data.
- Gold: Stores business-ready analytical data.

SQL command to create the required schemas.

````python
%sql
CREATE SCHEMA IF NOT EXISTS shadow_revenue_catalog.bronze;

CREATE SCHEMA IF NOT EXISTS shadow_revenue_catalog.silver;

CREATE SCHEMA IF NOT EXISTS shadow_revenue_catalog.gold;
````
